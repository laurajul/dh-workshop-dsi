# Welcome

This workshop is about finding and working with digitised cultural heritage on the web. Over three notebooks you will see how three ideas fit together:

1. **IIIF**: a shared standard that lets any viewer show images from any institution, at any size and detail.
2. **Europeana**: an aggregator that brings together descriptions of millions of objects from thousands of institutions, with an API you can query.
3. **CLIP**: a model that lets you search images by describing them in words.

At the end you will put your own selection of objects into a viewer.
